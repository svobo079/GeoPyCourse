```python
def temp_in_comfort_range(temp_celsius, temp_fahr_ideal=68.0, temp_fahr_range=5.0):
    temp_fahr = celsius_to_fahr(temp_celsius)
    temp_fahr_min = temp_fahr_ideal - temp_fahr_range
    temp_fahr_max = temp_fahr_ideal + temp_fahr_range
    temp_in_range = temp_fahr >= temp_fahr_min and temp_fahr <= temp_fahr_max
    return temp_in_range
```

```python
temp_in_comfort_range(21.0, 68.0, 5.0)
```

```python
temp_in_comfort_range(21, 45,2)
```

```python
def fahr_to_celsius(temp_fahr):
    """
    Converts temperature in Fahrenheit to degrees Celsius.

    Parameters
    ----------
    temp_fahr: <numerical>
        Temperature in Fahrenheit

    Returns
    -------
    <float>
        Converted temperature.
    """
    return ((temp_fahr - 32) * 5 / 9)
```

```python
print(fahr_to_celsius(32))
print(fahr_to_celsius(68))
print(fahr_to_celsius(91))
print(fahr_to_celsius(-17))
```

```python
def temp_converter(temp):
    if temp < -2:
        return(0)
    elif temp >= -2 and temp < 2:
        return(1)
    elif temp >= 2 and temp < 15:
        return(2)
    elif temp >= 15:
        return(3)
```
